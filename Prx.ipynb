```python
# sales = np.array([2500, 2700, 3000, 3200, 3100, 3300, 4000, 4200, 4100, 3900, 3700, 3600])
import numpy as np

sales = np.array([2500, 2700, 3000, 3200, 3100, 3300, 4000, 4200, 4100, 3900, 3700, 3600])

total_sales = np.sum(sales)
averge_sales = np.mean(sales)
high_sales_months = np.where(sales>4000)[0]+1 #Adding 1 to match month number

print(total_sales)
print(averge_sales)
print(high_sales_months)
```

```python
# IMG PROCESSING
image = np.array([
    [200, 202, 198, 201, 199],
    [203, 205, 204, 202, 200],
    [198, 197, 195, 199, 200],
    [201, 202, 204, 203, 202],
    [199, 198, 200, 201, 202]
])

normalized = (image - np.min(image)) / (np.max(image) - np.min(image))

threshold = np.where(image >= 200,1,0)

avg_brightness = np.mean(image)

print(threshold)
print(avg_brightness)
print(normalized)
```
