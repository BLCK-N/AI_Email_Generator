<!-- 1. Get all elements divisible by 5 -->
<!-- 2. Get all elements in the last column -->
<!-- 3. Get the indices of all elements > 20 -->
<!-- 4. Replace all values > 30 with 0 -->
<!-- 5. Extract a 2×2 submatrix from the top-left corner -->
<!-- 6. Get the row with the maximum sum -->
<!-- 7. Get the column with the minimum sum -->
<!-- 8. Count how many values are less than 25 -->
<!-- 9. Multiply every element in the second row by 2 -->
<!-- 10. Set the diagonal elements to 1 -->
<!-- 11. Transpose the array -->
<!-- 12. Flatten the array into 1D -->
13. Get only the even numbers
14. Create a mask of values greater than the column-wise mean
15. Set all negative values to zero (try after subtracting 20)
16. Get the element at (row=2, col=1)
17. Extract values from the border (no corners)
18. Calculate the average of elements in the third column
19. Sort the whole array as if it's 1D, then reshape to 3x3
20. Reverse each row

```python
import numpy as np 
arr = np.array([
    [10, 25, 5],
    [30, 15, 20],
    [35, 40, 45]
])

eles_by_5 = arr[arr % 5 == 0]
print("Elements divisible by 5:",eles_by_5)

eles_lst = arr[:,-1]
print("Elements in the last row", eles_lst)

indc_20 = np.argwhere(arr>20)
print("Indices greater than 20",indc_20)

replc_30 = np.where(arr > 30, 0, arr)
print("Replacing values greater than 30 with 0",replc_30)

submatrx_2x2 = arr[:2, :2]
print("Submatrix 2x2",submatrx_2x2);

max_row = arr[np.argmax(np.sum(arr, axis=1))]
print("Max row", max_row)

min_column = arr[:,np.argmin(np.sum(arr, axis = 0))]
print("Min column",min_column);

count_less_25 = np.sum(arr<25)
print("Less than 25",count_less_25)

secnd_row_tms2 = arr[1]*2
print("Second row multiplied by 2",secnd_row_tms2)

arr_dign_1 = arr.copy()
np.fill_diagonal(arr_dign_1,1)
print("Diagonal Elements to 1",arr_dign_1)

arr_trnsp = arr.T
print("Transpose",arr_trnsp)

flt_1d = arr.flatten()
print("Flattened array",flt_1d)

even = arr[arr % 2 == 0]
print("Even Numbers",even)

col_means = arr.mean(axis=0)
mask_col_mean = arr > col_means
print("Mask of values greater than column-wise mean",mask_col_mean)

arr_sub20 = arr - 20
arr_0 = np.where(arr_sub20 < 0,0,arr_sub20)
print("Negative values to 0",arr_0)

ele_2_1 = arr[2,1]
print("Element at row2 and col1",ele_2_1)

border_no_corner = np.array([arr[0,1], arr[1,0], arr[1,2], arr[2,1]])
print("Border Elements",border_no_corner)

avg_third_col = np.mean(arr[:, -2])
print("Avg of 3rd column",avg_third_col)

sort_to_3d = np.sort(arr, axis=None).reshape(3,3)
print("Sorted to 3d",sort_to_3d)

reverse_row = arr[:,::-1]
print("Reversed Row",reverse_row)
```
